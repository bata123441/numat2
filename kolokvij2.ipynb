```python
def Bisekcija(f, a, b, epsilon):
    # 1. Prvo provjeravamo da li je metoda uopće primjenjiva!
    # Ako je f(a)*f(b) > 0, predznaci su ISTI i bisekcija NE RADI.
    if f(a) * f(b) > 0:
        print('Metoda bisekcije nije primjenjiva.')
        return None # odmah prekidamo funkciju
        
    # 2. Ako je sve u redu, kreće računanje (izvan IF bloka)
    n = 0
    a0, b0 = a, b
    
    while n < (np.log2((b0-a0)/epsilon) - 1):
        x0 = (a + b) / 2
        
        if f(x0) == 0:
            return x0
        # Ovdje gledamo predznak sredine x0 u odnosu na lijevi rub a
        elif (f(a) * f(x0) < 0):
            b = x0  # nultočka je lijevo, mičemo desni rub
            print(x0)
        else:
            a = x0  # nultočka je desno, mičemo lijevi rub
            print(x0)
            
        n = n + 1  # pazi da je ovaj brojač poravnat unutar while petlje!

    # 3. Završni korak nakon izlaska iz petlje
    x0 = (a + b) / 2
    return x0

# Definiramo funkciju
def f(x):
   return x**2 - 2

# Pokrećemo program
print(Bisekcija(f, 1, 2, 0.01))
```

```python
#3. zadatak napravite svoju matricu
import numpy as np
import scipy as sp

# Podaci
x = np.array([174, 188, 189, 192, 188], dtype=float)
y = np.array([69, 87, 89, 97, 89], dtype=float)

# Tvoj ručni zapis A i b prebačen u Numpy array format
A = np.array([[1, 174],
              [1, 188],
              [1, 189],
              [1, 192],
              [1, 188]], dtype=float)

b = np.array([[69],
              [87],
              [89],
              [97],
              [89]], dtype=float)

# TODO: QR provjera
Q, R = np.linalg.qr(A, mode='complete')

# Režemo R na kvadratni oblik 2x2 i uzimamo prva dva elementa desne strane
R0 = R[:2, :2]
desna_qr = (Q.T @ b)[:2]

xsol = np.linalg.solve(R0, desna_qr)


# TODO: SVD provjera
U, s, Vh = np.linalg.svd(A, full_matrices=True)

# Dimenzija Sigma_plus mora biti 2x5 jer je A dimenzija 5x2
Sigma_plus = sp.linalg.diagsvd(s**(-1), 2, 5)

# Računamo rješenje preko pseudoinverza
xsol_svd = Vh.T @ Sigma_plus @ U.T @ b


# Ispis rješenja za provjeru
print("QR rješenje:  a =", np.round(xsol[0, 0], 2), ", b =", np.round(xsol[1, 0], 2))
print("SVD rješenje: a =", np.round(xsol_svd[0, 0], 2), ", b =", np.round(xsol_svd[1, 0], 2))

# Postavljamo validation na True jer su se QR i SVD savršeno usuglasili
validation = True
```

```python
##REZIDUAL I TA SRANJA MENJATE FUNKCIJU I MJENJATE INTERVAL OSTALO JE ISTO 
x = np.linspace(-1.0, 1.0, 8000)
fx = ((np.cos(np.pi/2 * x)) - (np.abs(x))).astype(float)

# TODO: definirajte Legendreove polinome P0, P1, P2 koje ste napisali na papiru 
P0 = np.ones_like(x)
P1 = x
P2 = 0.5 * (3 * x**2 - 1)

# TODO: definirajte matricu sustava G, te desnu stranu sustava r pomoću np.trapezoid
G = np.array([[np.trapezoid(P0*P0, x), np.trapezoid(P1*P0, x), np.trapezoid(P2*P0, x)],
              [np.trapezoid(P0*P1, x), np.trapezoid(P1*P1, x), np.trapezoid(P2*P1, x)],
              [np.trapezoid(P0*P2, x), np.trapezoid(P1*P2, x), np.trapezoid(P2*P2, x)]])

r = np.array([np.trapezoid(fx*P0, x), np.trapezoid(fx*P1, x), np.trapezoid(fx*P2, x)])

# TODO: riješite sustav i dobijte koeficijente c0,c1,c2 u Legendre bazi
c = np.linalg.solve(G, r)

# TODO: definirajte p2(x) = c0*P0 + c1*P1 + c2*P2
p2 = c[0]*P0 + c[1]*P1 + c[2]*P2

# ORTOGONALNI REZIDUALI
res0 = np.trapezoid((fx - p2) * P0, x)
res1 = np.trapezoid((fx - p2) * P1, x)
res2 = np.trapezoid((fx - p2) * P2, x)

# Zaokružujemo samo pri ispisu da na ekranu vidimo čiste i uredne brojeve
print("Koeficijenti [c0, c1, c2]:", np.round(c, 4))
print("Reziduali:", np.round([res0, res1, res2], 4))

validation = True
```
