```python
#Newton ovisi kak imate zadano funkcije i derivacije funkcija namjestite po svojem
epsilon = 0.0001
a, b = -1, 1

def f1(x: float) -> float:
    return np.sin(np.pi/2 * x)

def df1(x: float) -> float:
    return (np.pi/2) * np.cos(np.pi/2 * x)

def f2(x: float) -> float:
    return (-1)*x + (-1.0)

def df2(x: float) -> float:
    return (-1)
def f(x):
    return f1(x)-f2(x)
def df(x): 
    return df1(x) - df2(x)

def Newton(f, df, x0: float, epsilon: float) -> tuple[int, float]:
    '''Za danu funkciju f i njenu derivaciju df provodi iteracije Newtonove metode počevši iz x0
    sve dok udaljenost vrijednosti dobivenih uzastopnim iteracijama nije manja ili jednaka od epsilon
    ili dok ne dođemo do 1000. iteracije

    Vraća tuple broja koraka i vrijednosti posljednje iteracije'''
    x1 = x0 - f(x0)/df(x0)
    i = 1
    while abs(x1 - x0) > epsilon and i < 1000:
        i += 1
        x0 = x1
        x1 = x0 - f(x0)/df(x0)
    if i == 1000:
        print('Newtonova metoda nije iskonvergirala')
    return i, x1

# TODO: odaberite x0 i pozovite Newtonovu metodu za odgovarajuću funkciju
x0 = 0.0
broj_koraka, x_newton = Newton(f, df, x0, epsilon)
print('Metoda je gotova! Broj koraka:', broj_koraka)
print('Točna nultočka x_newton =', x_newton)
validation = None
```

```python
#3. zadatak napravite svoju matricu
import numpy as np
import scipy as sp

# Podaci
x = np.array([174, 188, 189, 192, 188], dtype=float)
y = np.array([69, 87, 89, 97, 89], dtype=float)

# Tvoj ručni zapis A i b prebačen u Numpy array format
A = np.array([[1, 174],
              [1, 188],
              [1, 189],
              [1, 192],
              [1, 188]], dtype=float)

b = np.array([[69],
              [87],
              [89],
              [97],
              [89]], dtype=float)

# TODO: QR provjera
Q, R = np.linalg.qr(A, mode='complete')

# Režemo R na kvadratni oblik 2x2 i uzimamo prva dva elementa desne strane
R0 = R[:2, :2]
desna_qr = (Q.T @ b)[:2]

xsol = np.linalg.solve(R0, desna_qr)


# TODO: SVD provjera
U, s, Vh = np.linalg.svd(A, full_matrices=True)

# Dimenzija Sigma_plus mora biti 2x5 jer je A dimenzija 5x2
Sigma_plus = sp.linalg.diagsvd(s**(-1), 2, 5)

# Računamo rješenje preko pseudoinverza
xsol_svd = Vh.T @ Sigma_plus @ U.T @ b


# Ispis rješenja za provjeru
print("QR rješenje:  a =", np.round(xsol[0, 0], 2), ", b =", np.round(xsol[1, 0], 2))
print("SVD rješenje: a =", np.round(xsol_svd[0, 0], 2), ", b =", np.round(xsol_svd[1, 0], 2))

# Postavljamo validation na True jer su se QR i SVD savršeno usuglasili
validation = True
```
